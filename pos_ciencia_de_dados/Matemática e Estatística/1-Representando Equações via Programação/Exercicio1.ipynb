### Exercício: Conversão de Temperatura

Crie uma função que converte a temperatura de Celsius para Fahrenheit e vice-versa. O usuário deve informar a temperatura e a escala de origem.

```python
esc, temp = input("escala (C ou F) e temperatura:").split()
temp = float(temp)
if esc == "C":
    result = temp* 9/5 + 32
elif esc == "F":
    result = (temp - 32) * 5/9
else:
    print('escala errada')
print(result)
```

### Exercício: Resolver uma Equação Linear

Faça um programa que resolve uma equação linear do tipo ax + b = 0. Os valores de a e b são fornecidos pelo usuário.

```python
a, b = input("A e B:").split()

a = float(a)
b = float(b)

if a == 0:
    print('Invalido')
else:
    x = -b/a
    print(x)
```

### Exercício: Conversão de Moedas

Escreva um programa que converte uma quantidade de uma moeda (como dólar) para outra (como euro), considerando uma taxa de câmbio fornecida pelo usuário.

```python
def conversao(valor: float, taxa: float):
    return valor * taxa

valor = float(input("valor:"))
taxa = float(input("taxa:"))
vf = conversao(valor, taxa)
print(f'Valor convertido: {vf}')
```

### Exercício: Verificar se um Número é Primo

Crie uma função que verifica se um número fornecido pelo usuário é primo ou não.

```python
def primo(x):
    if x<2:
        return False
    
    for i in range(2, int(x)):
        print(i)
        if x % i == 0:
            return False
    return True

x = int(input("numero:"))

print('Primo' if primo(x) else print('Não Primo'))
```
